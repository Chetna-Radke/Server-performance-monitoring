```python
import pandas as pd
import numpy as np


EXCEL_FILE = 'server_data.xlsx'
OUTPUT_FILE = 'cleaned_server_data.parquet'

print(f"--- 1. Data Ingestion from {EXCEL_FILE} ---")

# Load Excel sheets
metadata_df = pd.read_excel(EXCEL_FILE, sheet_name='Server_Metadata')
station1_df = pd.read_excel(EXCEL_FILE, sheet_name='Server_Performance_Station1')
station2_df = pd.read_excel(EXCEL_FILE, sheet_name='Server_Performance_Station2')

print("Data loaded successfully!")

# Combine logs
performance_df = pd.concat([station1_df, station2_df], ignore_index=True)

print("--- 2. Data Cleaning ---")

print("Columns:", performance_df.columns.tolist())   #to chcek the exact names of columns

# Drop unwanted columns
columns_to_drop = ['Config_Version', 'Last_Patch_Date', 'Deployment_Token']
performance_df = performance_df.drop(columns=columns_to_drop, errors='ignore')

# Drop rows with no Server_ID
performance_df.dropna(subset=['Server_ID'], inplace=True)


# Median imputations
performance_df['CPU_Utilization (%)'].fillna(performance_df['CPU_Utilization (%)'].median(), inplace=True)
performance_df['Memory_Usage (%)'].fillna(performance_df['Memory_Usage (%)'].median(), inplace=True)

print("--- 3. Data Enrichment (Join) ---")

final_df = pd.merge(
    performance_df,
    metadata_df,
    on='Server_ID',
    how='left'
)

# Make column names Power BI friendly
final_df.columns = (
    final_df.columns
    .str.replace(" ", "_")
    .str.replace("[^A-Za-z0-9_]+", "", regex=True)
)

final_df.to_parquet(OUTPUT_FILE, index=False)
print("SUCCESS! File saved:", OUTPUT_FILE)

print(final_df.head())
```
